```python

# Convert to DataFrame
df = pd.DataFrame(data)
```

```python
# ============================================
# STEP 3: Create basket matrix
# ============================================

basket = df["Items"].apply(
    lambda items: pd.Series(
        {item: 1 for item in items}
    )
).fillna(0)

basket.head()
```

```python
# ============================================
# STEP 4: Convert values to integers
# ============================================

basket = basket.astype(int)
```

```python
# We have 100 transactions
total_transactions = len(basket)

# Itemset must appear in at least 20 transactions
min_support_count = 20

print("\nTotal transactions:", total_transactions)
print("Minimum support count:", min_support_count)
```

```python
min_support = min_support_count / total_transactions

print("Minimum support:", min_support)
```

```python

# ============================================
# STEP 7: Apply Apriori
# ============================================

frequent_items = apriori(
    basket,
    min_support=min_support,
    use_colnames=True
)
```

```python

# Show 1-itemsets
one_itemsets = frequent_items[
    frequent_items['itemsets'].apply(len) == 1
]

print("1-Itemsets:")
print(one_itemsets)
```

```python
two_itemsets = frequent_items[
    frequent_items['itemsets'].apply(len) == 2
]

print("2-Itemsets:")
print(two_itemsets)
```

```python
three_itemsets = frequent_items[
    frequent_items['itemsets'].apply(len) == 3
]

print("3-Itemsets:")
print(three_itemsets)
```

```python

# ============================================
# STEP 9: Generate association rules
# ============================================

rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.50
)
```

```python

# ============================================
# STEP 10: Display association rules
# ============================================

print("\nAssociation Rules:")

print(
    rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ]
)

```

```python
# ============================================
# STEP 11: Sort rules by confidence
# ============================================

rules = rules.sort_values(
    by="confidence",
    ascending=False
)


print("\nTop Association Rules:")

print(
    rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(10)
)
```
